## 1. Load Image
* You can change the picture by changing the img variable

```python
import cv2, numpy as np
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = (5,5)

img = cv2.imread('pic1.png', cv2.IMREAD_GRAYSCALE)
if img is None:
    raise FileNotFoundError("Place pic.png next to the notebook.")
plt.imshow(img, cmap='gray'); plt.axis('off'); plt.title('Original Image'); plt.show()
```

## 2. Sobel Filters


```python
sx = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
sy = np.array([[-1,-2,-1],[0,0,0],[1,2,1]])

plt.imshow(cv2.filter2D(img,-1,sx), cmap='gray'); plt.axis('off'); plt.title('Sobel X'); plt.show()
plt.imshow(cv2.filter2D(img,-1,sy), cmap='gray'); plt.axis('off'); plt.title('Sobel Y'); plt.show()
```

## 3. Prewitt Filters

```python
px = np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
py = np.array([[-1,-1,-1],[0,0,0],[1,1,1]])

plt.imshow(cv2.filter2D(img,-1,px), cmap='gray'); plt.axis('off'); plt.title('Prewitt X'); plt.show()
plt.imshow(cv2.filter2D(img,-1,py), cmap='gray'); plt.axis('off'); plt.title('Prewitt Y'); plt.show()
```

## 4. Laplacian

```python
L = np.array([[0,-1,0],[-1,4,-1],[0,-1,0]])
plt.imshow(cv2.filter2D(img,-1,L), cmap='gray'); plt.axis('off'); plt.title('Laplacian'); plt.show()
```

## 5. Sharpen

```python
sh = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
plt.imshow(cv2.filter2D(img,-1,sh), cmap='gray'); plt.axis('off'); plt.title('Sharpen'); plt.show()
```

## 6. Emboss

```python
emb = np.array([[-2,-1,0],[-1,1,1],[0,1,2]])
plt.imshow(cv2.filter2D(img,-1,emb), cmap='gray'); plt.axis('off'); plt.title('Emboss'); plt.show()
```

## 7. Box Blur

```python
box = (1/9)*np.ones((3,3))
plt.imshow(cv2.filter2D(img,-1,box), cmap='gray'); plt.axis('off'); plt.title('Box Blur'); plt.show()
```

## 8. Gaussian Blur

```python
gauss = (1/16)*np.array([[1,2,1],[2,4,2],[1,2,1]])
plt.imshow(cv2.filter2D(img,-1,gauss), cmap='gray'); plt.axis('off'); plt.title('Gaussian Blur'); plt.show()
```
